## use tiktoken when you need to know how much text you are sending to a model, not merely how many words or characters it contains.

tiktoken is a fast tokenizer used to convert text into the tokens processed by OpenAI models.

Count tokens:

```python
import tiktoken

text = "Data science and machine learning are useful."

encoding = tiktoken.get_encoding("o200k_base")
tokens = encoding.encode(text)

print("Tokens:", tokens)
print("Token count:", len(tokens))
```

select the encoding associated with a particular model:

```python
import tiktoken

text = "Explain machine learning in simple words."

encoding = tiktoken.encoding_for_model("gpt-4o")
token_count = len(encoding.encode(text))

print(f"Token count: {token_count}")
```

A common document-chunking check looks like this:

```python
def count_tokens(text: str) -> int:
    encoding = tiktoken.get_encoding("o200k_base")
    return len(encoding.encode(text))


document = "A very long document..."

if count_tokens(document) > 4:
    print("The document needs to be divided into smaller chunks.")
```
